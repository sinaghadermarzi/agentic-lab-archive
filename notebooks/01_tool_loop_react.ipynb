# 01 — Tool-use loop / ReAct

The foundational loop is observe → decide → act → observe. The policy below is deterministic so the trace is inspectable; replace it with an LLM decision boundary only after the tool contract and step budget are clear.

Primary reading: [ReAct](https://arxiv.org/abs/2210.03629).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'notebooks'))
from lab_core.patterns import run_tool_loop

facts = {'python': 'Python was first released in 1991.', 'react': 'ReAct interleaves reasoning traces with actions.'}

def lookup(term):
    return facts.get(str(term).lower(), 'not found')

def choose_action(goal, trace):
    if not trace:
        return {'tool': 'lookup', 'input': 'react'}
    return {'final': f"Goal: {goal}. Evidence: {trace[-1]['observation']}"}

result = run_tool_loop('Explain the ReAct pattern', choose_action, {'lookup': lookup}, max_steps=3)
print(result.answer)
print(result.trace)
```
