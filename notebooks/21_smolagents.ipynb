# 21 — Hugging Face smolagents

smolagents keeps the agent loop deliberately small (~1k lines of core logic) and makes one distinctive bet: a `CodeAgent` writes its actions as Python snippets instead of JSON tool calls, which the project reports needs fewer steps on hard benchmarks. A conventional `ToolCallingAgent` is also provided, and model classes (`InferenceClientModel`, `LiteLLMModel`, `OpenAIModel`, `TransformersModel`, ...) keep it provider-agnostic.

The code-as-action bet sharpens a boundary this lab keeps returning to: when the model's *decision* is executable code, the sandbox becomes the permission boundary (notebook 00). smolagents supports sandboxed executors (Docker, E2B, and others) for exactly this reason — run untrusted generated code the way you would any untrusted input (notebook 05).

In lab terms, this is the notebook-01 loop with the decide step emitting code instead of a tool-name/arguments pair.

Primary sources: [documentation](https://huggingface.co/docs/smolagents) and [repository](https://github.com/huggingface/smolagents).

**Setup:** `pip install -e '.[smolagents]'` (plus an HF token or provider key for live runs; `pip install -e '.[all]'` installs every extra at once).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'notebooks'))
from lab_core.frameworks import FRAMEWORKS, availability

print(FRAMEWORKS['smolagents']['source'])
if not availability().get('smolagents'):
    print('Install the optional extra to instantiate this example.')

official_shape = '''
from smolagents import CodeAgent, InferenceClientModel, tool

@tool
def get_forecast(city: str) -> str:
    """Deterministic forecast for a known city.

    Args:
        city: City name to look up.
    """
    return {'oslo': 'Snow, -3C', 'lisbon': 'Sunny, 21C'}.get(city.lower(), 'unknown city')

agent = CodeAgent(tools=[get_forecast], model=InferenceClientModel())
# agent.run('What is the weather in Oslo?')  # requires an HF token or another provider
'''
print(official_shape)
```
