# 26 — LangChain deepagents

The `deepagents` package ships the four pillars as defaults on top of LangGraph's `create_agent`: a `write_todos` planning tool, filesystem tools over pluggable backends (local, sandboxed, or remote), a `task` tool that spawns sub-agents with isolated context, and a detailed default system prompt plus loadable skills — the same architecture this track built by hand, which is why the mapping is row-for-row:

| Pillar | Lab toy (`lab_core.deep`) | `deepagents` | Claude Code |
| --- | --- | --- | --- |
| plan artifact | `write_todos` / `TodoList` | `write_todos` | todo list |
| context offload | `write_file`/`read_file`/`ls` on `VirtualFS` | filesystem tools + backends | file tools |
| isolated delegation | `task` → `SubAgent` | `task` → subagents | Task tool (subagents) |
| prompt & skills | `compose_system_prompt` + `load_skill` | system prompt + skills | skills |

Unlike notebooks 17 and 19, there is no keyless live cell here: a deep agent's policy *is* an LLM, so the example below is the official shape only (Track B convention).

Primary sources: [repository](https://github.com/langchain-ai/deepagents), [documentation](https://docs.langchain.com/oss/python/deepagents/overview), and the [Deep Agents post](https://www.langchain.com/blog/deep-agents).

**Design translation.** `create_deep_agent` hands you the harness — plan, files, sub-agents, skills, plus checkpointing via LangGraph. What it does not solve remains your architecture: whether the plan the model writes is any good (notebook 03's verification), which delegations to trust (04), and where a human must approve (05). The pillars transfer to any stack — including the frameworks of Track B — because they are context disciplines, not framework features.
