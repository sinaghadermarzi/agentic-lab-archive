# 03 — Verification and bounded adversarial review

A useful critic has an explicit rubric and access to evidence. The loop below revises a candidate only when the verifier reports a concrete issue.

Primary reading: [Reflexion](https://arxiv.org/abs/2303.11366) and [Anthropic — Building Effective AI Agents](https://resources.anthropic.com/building-effective-ai-agents).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'notebooks'))
from lab_core.patterns import evaluate_optimize

def evaluate(candidate):
    ok = 'source:' in candidate and len(candidate) >= 30
    return {'pass': ok, 'issues': [] if ok else ['add evidence and a fuller explanation']}

def revise(candidate, report):
    return candidate + ' source: primary documentation'

final, reports = evaluate_optimize('A short draft.', evaluate, revise, max_rounds=2)
print(final)
print(reports)
```
