# 04 — Delegation, agent-as-tool, and handoffs

Agent-as-tool keeps the supervisor in control of the final answer. A handoff transfers ownership. Both are useful; the failure mode is ambiguous ownership and duplicated work.

Compare the official [OpenAI Agents SDK orchestration guidance](https://openai.github.io/openai-agents-python/).

```python
def researcher(task):
    return {'worker': 'researcher', 'artifact': f'evidence for: {task}'}

def reviewer(task):
    return {'worker': 'reviewer', 'artifact': f'risks and gaps in: {task}'}

def supervisor(task):
    workers = [researcher(task), reviewer(task)]
    return {'owner': 'supervisor', 'artifacts': workers, 'final': f'synthesized answer for: {task}'}

print(supervisor('whether a draft is ready to publish'))
print('Handoff variant: after routing, the specialist owns the next user-facing turn.')
```
