# 11 — Microsoft Agent Framework

Microsoft Agent Framework combines agent abstractions with enterprise-oriented workflows. Its orchestration families include sequential, concurrent, handoff, group chat, and Magentic patterns; approval-required tools make authority boundaries explicit.

**Lineage: AutoGen and Semantic Kernel.** Agent Framework is Microsoft's official convergence of its two earlier frameworks. **AutoGen** (multi-agent conversations, group chat) and **Semantic Kernel** (kernel functions, plugins, planners) are both in maintenance mode — bug fixes and security patches only, no new features — and Microsoft's migration guides map their concepts onto Agent Framework: AutoGen's `AssistantAgent` → `ChatAgent`, `FunctionTool` → `@ai_function`-style tools, event-driven group chats → workflow orchestrations; Semantic Kernel's `KernelFunction`s convert via `.as_agent_framework_tool()`. If you arrive here from an AutoGen or Semantic Kernel codebase, this lesson is where those concepts now live. (A separately redesigned community line also continues as [AG2](https://github.com/ag2ai/ag2).)

Migration guides: [from AutoGen](https://learn.microsoft.com/en-us/agent-framework/migration-guide/from-autogen/) and [from Semantic Kernel](https://learn.microsoft.com/en-us/agent-framework/migration-guide/from-semantic-kernel/).

Primary sources: [documentation](https://learn.microsoft.com/en-us/agent-framework/) and [workflow orchestrations](https://learn.microsoft.com/en-us/agent-framework/workflows/orchestrations/).
