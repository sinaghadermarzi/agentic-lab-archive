# 05 — State, human gates, and budgets

Durable systems persist meaningful state, pause at authority boundaries, and terminate on observable conditions. This notebook demonstrates those boundaries without a framework.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'notebooks'))
from lab_core.patterns import ApprovalGate, Budget, CheckpointStore

budget = Budget(max_steps=3, max_retries=1, max_spend=0.10)
store = CheckpointStore()
state = {'status': 'drafted', 'artifact': 'synthetic report'}
budget.charge(spend=0.01)
store.save('run-1', state)

gate = ApprovalGate(lambda proposal: proposal['risk'] == 'low')
approved = gate.require({'action': 'save report', 'risk': 'low'})
print('resumed state:', store.load('run-1'))
print('approved action:', approved)
print('budget:', budget)
```
