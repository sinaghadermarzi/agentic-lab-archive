# 07 — LangGraph

LangGraph expresses stateful agent workflows as graphs with explicit nodes and edges. Use it when checkpointing, cycles, routing, and inspectable control flow matter.

Primary source: [LangGraph Python install](https://docs.langchain.com/oss/python/langgraph/install).

**Setup:** `pip install -e '.[langgraph]'` (plus an OpenAI-compatible key for live cells; `pip install -e '.[all]'` installs every extra at once).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'notebooks'))
from lab_core.frameworks import FRAMEWORKS, availability

print(FRAMEWORKS['LangGraph']['source'])
if not availability().get('LangGraph'):
    print('Optional package not installed; showing the official API shape only.')

official_shape = '''
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END

class State(TypedDict):
    text: str

def normalize(state):
    return {'text': state['text'].strip().lower()}

builder = StateGraph(State)
builder.add_node('normalize', normalize)
builder.add_edge(START, 'normalize')
builder.add_edge('normalize', END)
graph = builder.compile()
print(graph.invoke({'text': ' Hello ' }))
'''
print(official_shape)
```
