# 20 — The agent-protocol landscape

Notebooks 16–19 covered the two protocols with hands-on weight today: **MCP** for the tool boundary and **A2A** for the peer boundary. Around them sits a wider, fast-moving landscape. The useful mental model is a stack of boundaries, each owned by at most one or two credible standards:

| Boundary | Standard(s) | Covered |
| --- | --- | --- |
| agent ↔ tools and context | MCP | notebooks 16–17 |
| agent ↔ agent | A2A (+ ANP for decentralized/open networks) | notebooks 18–19 |
| agent ↔ user-facing app | AG-UI | this survey |
| agent ↔ payments | AP2 | this survey |
| identity, discovery, messaging infra | AGNTCY components (OASF, directory, SLIM) | this survey |
| code editor ↔ coding agent | Agent Client Protocol | this survey |

The survey entries stop at concepts deliberately: AG-UI is TypeScript/frontend-centric (this lab is Python-first), AP2 and ANP are still emerging, AGNTCY is infrastructure rather than application architecture, and the Agent Client Protocol serves editor tooling.
