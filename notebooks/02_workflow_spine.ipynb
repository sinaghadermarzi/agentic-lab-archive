# 02 — Deterministic workflow spine

Keep predictable control flow in code and reserve model judgment for interpretation, classification, and synthesis. This example combines routing with parallel fan-out/fan-in.

Primary reading: [Anthropic workflow patterns](https://resources.anthropic.com/building-effective-ai-agents).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'notebooks'))
from lab_core.patterns import fan_out_fan_in, route

def classify(request):
    return 'technical' if any(word in request.lower() for word in ('api', 'code', 'python')) else 'general'

branches = {
    'technical': lambda text: {'route': 'technical', 'next': 'inspect code and run tests'},
    'general': lambda text: {'route': 'general', 'next': 'retrieve policy and draft response'},
}
print(route('How do I test a Python API?', classify, branches))

sources = ['official docs', 'unit tests', 'change log']
evidence = fan_out_fan_in(sources, lambda source: f'checked {source}', lambda rows: {'evidence': rows})
print(evidence)
```
