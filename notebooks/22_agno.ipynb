# 22 — Agno

Agno pairs the familiar building blocks — `Agent`, `Team` — with an opinionated runtime: **AgentOS**, a control plane that serves agents as services with sessions, memory, and management UI in your own infrastructure. Choose it when the operational boundary (running and observing many agents) matters as much as authoring them.

In lab terms: a `Team` is notebook 04's supervisor–worker with the collaboration mode picked by configuration, and AgentOS is the operational envelope that notebooks 00 and 05 build by hand — budgets, state, and human oversight as a hosted runtime.

Primary sources: [documentation](https://docs.agno.com/) and [repository](https://github.com/agno-agi/agno).
